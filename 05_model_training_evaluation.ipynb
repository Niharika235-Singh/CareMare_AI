```python
import pandas as pd
import os

# ==============================
# Load DDXPlus Train Dataset
# ==============================

train = pd.read_csv(os.path.join(chandashekar8_ddxplus_path, "train.csv"))

TARGET = "PATHOLOGY"

sampled = []

for disease, group in train.groupby(TARGET):

    n = len(group)

    # ---------- Sampling Policy ----------

    if n <= 5000:
        take = n

    elif n <= 10000:
        take = 350

    elif n <= 20000:
        take = 550

    elif n <= 30000:
        take = 650

    else:
        take = 750

    sampled.append(
        group.sample(
            n=min(take, n),
            random_state=42
        )
    )

# Merge all diseases
sampled_df = pd.concat(sampled).reset_index(drop=True)

print("="*50)
print("Final Dataset Shape")
print(sampled_df.shape)

print("="*50)
print(sampled_df["PATHOLOGY"].value_counts())

sampled_df.to_csv("DDXPlus_22K.csv", index=False)

print("Saved Successfully!")
```

```python
def build_evidence_vocab(evidences_data):

    # Special Tokens
    vocab = {
        "<PAD>": 0,
        "<UNK>": 1
    }

    idx = 2

    # Har evidence par loop
    for code, info in evidences_data.items():

        # Evidence type
        data_type = info.get("data_type", "B")

        # -----------------------------
        # Binary Evidence
        # -----------------------------
        if data_type == "B":

            vocab[code] = idx
            idx += 1

        # -----------------------------
        # Categorical / Multi-choice
        # -----------------------------
        else:

            possible_values = info.get("possible-values", {})

            for value in possible_values:

                token = f"{code}_@_{value}"

                vocab[token] = idx

                idx += 1

    return vocab
```

```python
evidence_vocab = build_evidence_vocab(evidence_data)
```

```python
import ast

def convert_evidences_to_ids(evidence_string, vocab):

    # String → List
    evidence_list = ast.literal_eval(evidence_string)

    # Token → ID
    evidence_ids = []

    for token in evidence_list:

        evidence_ids.append(
            vocab.get(token, vocab["<UNK>"])
        )

    return evidence_ids
```

```python
import torch
from torch.utils.data import Dataset

class DDXDataset(Dataset):
    def __init__(self, data):
        self.sequences = data["SEQUENCE"].tolist()
        self.ages = torch.tensor(data["AGE"].values, dtype=torch.float32)
        self.sexes = torch.tensor(data["SEX"].values, dtype=torch.long)
        self.labels = torch.tensor(data["PATHOLOGY"].values, dtype=torch.long)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        sequence = torch.tensor(self.sequences[idx], dtype=torch.long)
        age = self.ages[idx]
        sex = self.sexes[idx]
        label = self.labels[idx]
        return sequence, age, sex, label

train_dataset = DDXDataset(train_data)
val_dataset = DDXDataset(val_data)
test_dataset = DDXDataset(test_data)
```

```python
from torch.nn.utils.rnn import pad_sequence

def collate_fn(batch):

    sequences, ages, sexes, labels = zip(*batch)

    lengths = torch.tensor(
        [len(seq) for seq in sequences],
        dtype=torch.long
    )

    sequences = pad_sequence(
        sequences,
        batch_first=True,
        padding_value=0
    )

    ages = torch.stack(ages)

    sexes = torch.stack(sexes)

    labels = torch.stack(labels)

    return sequences, lengths, ages, sexes, labels
```

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    collate_fn=collate_fn
)

val_loader = DataLoader(
    val_dataset,
    batch_size=64,
    shuffle=False,
    collate_fn=collate_fn
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False,
    collate_fn=collate_fn
)
```

```python
import torch
import torch.nn as nn

class DiagnosisModel(nn.Module):

    def __init__(
        self,
        vocab_size,
        embedding_dim,
        hidden_dim,
        num_layers,
        num_classes,
        dropout=0.3
    ):

        super(DiagnosisModel, self).__init__()

        # -----------------------------
        # Embedding Layer
        # -----------------------------
        self.embedding = nn.Embedding(
            num_embeddings=vocab_size,
            embedding_dim=embedding_dim,
            padding_idx=0
        )

        # -----------------------------
        # LSTM Layer
        # -----------------------------
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0,
            bidirectional=False
        )

        # -----------------------------
        # Dropout Layer
        # -----------------------------
        self.dropout = nn.Dropout(dropout)

        # -----------------------------
        # Final Classifier
        # -----------------------------
        self.fc = nn.Linear(
            hidden_dim + 2,
            num_classes
        )

    def forward(self, sequence, lengths, age, sex):

        # -----------------------------
        # Embedding
        # -----------------------------
        embedded = self.embedding(sequence)

        # -----------------------------
        # Pack padded sequences
        # -----------------------------
        packed = nn.utils.rnn.pack_padded_sequence(
            embedded,
            lengths.cpu(),
            batch_first=True,
            enforce_sorted=False
        )

        # -----------------------------
        # LSTM
        # -----------------------------
        packed_output, (hidden, cell) = self.lstm(packed)

        # -----------------------------
        # Last Hidden State
        # -----------------------------
        hidden = hidden[-1]

        # -----------------------------
        # Dropout
        # -----------------------------
        hidden = self.dropout(hidden)

        # -----------------------------
        # Concatenate Age & Sex
        # -----------------------------
        age = age.unsqueeze(1)
        sex = sex.unsqueeze(1)

        features = torch.cat(
            (hidden, age, sex),
            dim=1
        )

        # -----------------------------
        # Final Classification
        # -----------------------------
        output = self.fc(features)

        return output
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print(device)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = DiagnosisModel(
    vocab_size=len(evidence_vocab),
    embedding_dim=128,
    hidden_dim=128,
    num_layers=2,
    num_classes=49,
    dropout=0.3
)

model = model.to(device)
```

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

scaler = torch.cuda.amp.GradScaler()
```

```python
# ==========================================
# Training + Validation (Single Loop)
# ==========================================

NUM_EPOCHS = 20
best_val_loss = float("inf")

train_losses = []
val_losses = []

train_accs = []
val_accs = []

for epoch in range(NUM_EPOCHS):

    # =====================================
    # TRAINING
    # =====================================
    model.train()

    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for sequences, lengths, ages, sexes, labels in train_loader:

        sequences = sequences.to(device)
        lengths = lengths.to(device)
        ages = ages.to(device)
        sexes = sexes.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        with torch.cuda.amp.autocast():

            outputs = model(
                sequences,
                lengths,
                ages,
                sexes
            )

            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        train_loss += loss.item()

        predictions = outputs.argmax(dim=1)

        train_correct += (predictions == labels).sum().item()
        train_total += labels.size(0)

    train_loss /= len(train_loader)
    train_acc = 100 * train_correct / train_total

    train_losses.append(train_loss)
    train_accs.append(train_acc)

    # =====================================
    # VALIDATION
    # =====================================
    model.eval()

    val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for sequences, lengths, ages, sexes, labels in val_loader:

            sequences = sequences.to(device)
            lengths = lengths.to(device)
            ages = ages.to(device)
            sexes = sexes.to(device)
            labels = labels.to(device)

            outputs = model(
                sequences,
                lengths,
                ages,
                sexes
            )

            loss = criterion(outputs, labels)

            val_loss += loss.item()

            predictions = outputs.argmax(dim=1)

            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

    val_loss /= len(val_loader)
    val_acc = 100 * val_correct / val_total

    val_losses.append(val_loss)
    val_accs.append(val_acc)

    # =====================================
    # PRINT RESULT
    # =====================================
    print("=" * 70)
    print(f"Epoch [{epoch+1}/{NUM_EPOCHS}]")

    print(f"Train Loss      : {train_loss:.4f}")
    print(f"Train Accuracy  : {train_acc:.2f}%")

    print(f"Validation Loss : {val_loss:.4f}")
    print(f"Validation Acc  : {val_acc:.2f}%")

    # =====================================
    # SAVE BEST MODEL
    # =====================================
    if val_loss < best_val_loss:

        best_val_loss = val_loss

        torch.save(model.state_dict(), "best_model.pth")

        print(" Best Model Saved")

print("\n Training Completed Successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

def evaluate_model(model, loader, device):

    model.eval()

    all_predictions = []
    all_labels = []

    with torch.no_grad():

        for sequences, lengths, ages, sexes, labels in loader:

            sequences = sequences.to(device)
            lengths = lengths.to(device)
            ages = ages.to(device)
            sexes = sexes.to(device)

            outputs = model(
                sequences,
                lengths,
                ages,
                sexes
            )

            predictions = torch.argmax(outputs, dim=1)

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.numpy()
            )

    return all_labels, all_predictions
```

```python
true_labels, predicted_labels = evaluate_model(
    model,
    test_loader,
    device
)
```
